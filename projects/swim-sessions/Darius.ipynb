```python
# time_value = '41.50'
time_value = '1:27.95'

def time_to_hundredths_seconds(time_value):
    parts = time_value.split(":")
    if len(parts) == 2:
        minutes, rest = parts
    else:
        minutes = 0
        rest = parts[0]

    parts = rest.split(".")
    if len(parts) == 2:
        seconds, hundredths = parts
    else:
        seconds = parts[0]
        hundredths = 0

    total_hundredths = (int(minutes) * 60 * 100) + (int(seconds) * 100) + int(hundredths)
    return total_hundredths

time_to_hundredths_seconds(time_value)
```

```python
times
```

```python
for t in times:
    print(t, time_to_hundredths_seconds(t))
```

```python
time_values = []

for t in times:
    value = time_to_hundredths_seconds(t)
    time_values.append(value)
    print(f'{t} -> {value}')
```

```python
def time_to_minutes_seconds_hundredths(value):
    mins_secs, hundredths = str(round(value / 100, 2)).split(".")
    mins_secs = int(mins_secs)
    mins = mins_secs // 60
    secs = mins_secs - (mins * 60)
    return f'{mins}:{secs}.{hundredths}'

time_to_minutes_seconds_hundredths(8795.5)
```
