```python
for n, table in enumerate(tables):
    rows = table.find("tr", mode="all")
    cols = rows[-1].find("td", mode="all")
    print(f"Table #{n}, {len(rows)} rows, {len(cols)} columns\n")
```

```python
table = tables[0]

long_course_men_records = {}
for row in table.find("tr", mode="all")[1:]:
    columns = row.find("td", mode="all")
    event = columns[0].text
    time = columns[1].text
    name = columns[3].text
    country = columns[4].text
    date = columns[5].text
    long_course_men_records[event] = {
        "time": time,
        "name": name,
        "country": country,
        "date": date
    }

print(long_course_men_records)
len(long_course_men_records)
```

```python
TABLE_RECORDS = (0, 1, 3, 4)
COURSES = ("LC Men", "LC Women", "SC Men", "SC Women")
world_records = {}
for table_record, course in zip(TABLE_RECORDS, COURSES):
    world_records[course] = {}
    for row in tables[table_record].find("tr", mode="all")[1:]:
        columns = row.find("td", mode="all")
        event = columns[0].text
        time = columns[1].text
        name = columns[3].text
        country = columns[4].text
        date = columns[5].text
        world_records[course][event] = {
            "time": time,
            "name": name,
            "country": country,
            "date": date
        }
print(world_records)
```
