```python
import os

import swimclub

data_files = os.listdir(swimclub.DATA_FOLDER)

if ".DS_Store" in data_files:
    data_files.remove(".DS_Store")

swimmers = []

for n, file in enumerate(data_files):
    swimmers.append((data_files[n].split("-", 1)[0], file))

print(swimmers)
len(swimmers)
```

```python
unique_swimmers = []
for n, swimmer in enumerate(swimmers):
    if swimmer[0] not in unique_swimmers:
        unique_swimmers.append(swimmer[0])

print(unique_swimmers)
len(unique_swimmers)
```

```python
swimmers_dict = {}

for data_file in data_files:
    swimmer = data_file.split("-", 1)[0]

    if swimmer not in swimmers_dict:
        swimmers_dict[swimmer] = []

    swimmers_dict[swimmer].append(data_file)

print(swimmers_dict)

print(len(swimmers_dict))
```
