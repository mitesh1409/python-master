```python
import os

import swimclub

data_files = os.listdir(swimclub.DATA_FOLDER)

if ".DS_Store" in data_files:
    data_files.remove(".DS_Store")

swimmers = []

for n, file in enumerate(data_files):
    swimmers.append((data_files[n].split("-", 1)[0], file))

print(swimmers)
len(swimmers)
```

```python
unique_swimmers = []
for n, swimmer in enumerate(swimmers):
    if swimmer[0] not in unique_swimmers:
        unique_swimmers.append(swimmer[0])

print(unique_swimmers)
len(unique_swimmers)
```
